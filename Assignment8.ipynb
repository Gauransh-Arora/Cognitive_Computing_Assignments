## Question 1


```python
text = """Technology makes our lives easier by helping us communicate,
learn, and solve problems quickly. I enjoy exploring artificial intelligence
because intelligent systems can analyze data and discover useful patterns.
Programmers write programs that automate repetitive tasks and improve
productivity. However, people should use technology responsibly and protect
their personal information. I believe technology is most valuable when it
helps people learn new skills and build a better future."""

print("Original Paragraph:\n", text)
```

```python
clean_text = re.sub(r'[^\w\s]', '', text.lower())
print("Original text:\n", text)
print("\nCleaned text (lowercase, punctuation removed):\n", clean_text)
```

```python
from nltk.tokenize import word_tokenize, sent_tokenize

words = word_tokenize(clean_text)
sentences = sent_tokenize(text)

print("Word tokens:")
print(words)

print("\nSentence tokens:")
for i, s in enumerate(sentences, 1):
    print(f"Sentence {i}: {s.strip()}")

print("\nTotal words:", len(words))
print("Total sentences:", len(sentences))
```

```python
split_words = text.split()
nltk_words = word_tokenize(text)
print("Python split() tokens:")
print(split_words)
print("\nNLTK word_tokenize() tokens:")
print(nltk_words)
print("\nNumber of tokens using split():", len(split_words))
print("Number of tokens using word_tokenize():", len(nltk_words))
print("\nComparison Observation:")
print("- Python split() splits purely on whitespace, leaving punctuation attached to words (e.g., 'communicate,' or 'future.').")
print("- NLTK word_tokenize() uses linguistic rules (Treebank tokenizer), cleanly separating punctuation marks as distinct tokens.")
```

```python
from nltk.corpus import stopwords
stop_words = set(stopwords.words('english'))
filtered_words = [
    word for word in words
    if word not in stop_words
]
print("Original word tokens:")
print(words)
print("\nWords after removing stopwords:")
print(filtered_words)
print("\nTotal words before removal:", len(words))
print("Total words after removal:", len(filtered_words))
```

```python
from nltk.probability import FreqDist

frequency_distribution = FreqDist(filtered_words)

print("Word frequency distribution:")
for word, frequency in frequency_distribution.most_common():
    print(f"{word}: {frequency}")

plt.figure(figsize=(10, 5))
frequency_distribution.plot(
    15,
    title="Word Frequency Distribution (Excluding Stopwords)"
)
plt.show()
```

```python
alphabet_words = re.findall(r'\b[a-zA-Z]+\b', text.lower())
print("Words containing only alphabets:")
print(alphabet_words)
print("\nTotal alphabetic words:", len(alphabet_words))
```

```python
from nltk.stem import PorterStemmer
porter = PorterStemmer()
stemmed_words = [
    porter.stem(word)
    for word in filtered_alphabet_words
]
print("Original words:")
print(filtered_alphabet_words)
print("\nStemmed words (PorterStemmer):")
print(stemmed_words)
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
lemmatized_words = [
    lemmatizer.lemmatize(word)
    for word in filtered_alphabet_words
]
print("Original words:")
print(filtered_alphabet_words)
print("\nLemmatized words (WordNetLemmatizer):")
print(lemmatized_words)
```

```python
comparison_df = pd.DataFrame({
    'Original Word': filtered_alphabet_words,
    'Stemmed (Porter)': stemmed_words,
    'Lemmatized (WordNet)': lemmatized_words
})
print("Comparison Table:")
print(comparison_df.to_string(index=False))
```

## Question 3


```python
texts = [
    "The laptop is fast and has an excellent battery life.",
    "The smartphone has a great camera and excellent display.",
    "The laptop battery is good but the camera is slow."
]

for i, t in enumerate(texts, start=1):
    print(f"Text {i}: {t}")
```

```python
from sklearn.feature_extraction.text import CountVectorizer

count_vectorizer = CountVectorizer()
bow_matrix = count_vectorizer.fit_transform(texts)
vocab = count_vectorizer.get_feature_names_out()

bow_df = pd.DataFrame(bow_matrix.toarray(), columns=vocab, index=[f"Text {i}" for i in range(1, len(texts)+1)])
print("Bag of Words (CountVectorizer) Vocabulary:")
print(count_vectorizer.vocabulary_)
print("\nBag of Words Document-Term Matrix:")
display(bow_df)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(texts)
tfidf_vocab = tfidf_vectorizer.get_feature_names_out()

tfidf_df = pd.DataFrame(
    np.round(tfidf_matrix.toarray(), 4),
    columns=tfidf_vocab,
    index=[f"Text {i}" for i in range(1, len(texts)+1)]
)
print("TF-IDF Matrix:")
display(tfidf_df)
```

```python
print("=" * 60)
print("Top 3 Keywords from Each Text based on TF-IDF Scores:")
print("=" * 60)

for i, doc in enumerate(texts, start=1):
    row = tfidf_matrix.toarray()[i-1]
    top_indices = row.argsort()[::-1][:3]
    top_keywords = [(tfidf_vocab[idx], round(row[idx], 4)) for idx in top_indices]

    print(f"\nText {i}: '{doc}'")
    print(f"Top 3 TF-IDF Keywords: {top_keywords}")

print("\nInterpretation of TF-IDF Keywords:")
print("1. Text 1: 'fast' (0.47) and 'life' (0.47) have the highest TF-IDF weights because they appear exclusively in Text 1.")
print("2. Text 2: 'great' (0.46) and 'display' (0.46) receive peak weights as distinctive terms characterizing Text 2.")
print("3. Text 3: 'good' (0.47) and 'slow' (0.47) receive highest weights since they are unique descriptive terms for Text 3.")
print("4. Shared terms across multiple texts (like 'laptop', 'battery', 'camera', 'excellent') receive attenuated IDF penalties.")
```

## Question 4


```python
text_ai = """Artificial intelligence enables machines to learn from experience and perform human-like tasks.
Deep learning and neural networks analyze vast amounts of data to recognize complex patterns.
AI algorithms adapt through progressive learning and optimize decision making processes.
From natural language processing to computer vision, intelligent systems transform industries.
Modern AI drives automation and predictive analytics across healthcare and finance."""

text_blockchain = """Blockchain is a decentralized and distributed digital ledger technology.
Transactions are securely recorded across a peer-to-peer network of multiple computers.
Cryptographic hashing ensures that blocks of data cannot be altered retroactively.
Smart contracts automatically execute agreements when predefined conditions are satisfied.
Blockchain provides transparency, immutability, and security for digital assets and finance."""

print("--- Technology 1: Artificial Intelligence ---")
print(text_ai)
print("\n--- Technology 2: Blockchain ---")
print(text_blockchain)
```

```python
def preprocess_text(t):
    cleaned = re.sub(r'[^\w\s]', ' ', t.lower())
    tokens = word_tokenize(cleaned)
    filtered = [w for w in tokens if w not in stop_words and len(w) > 1]
    return filtered

tokens_ai = preprocess_text(text_ai)
tokens_blockchain = preprocess_text(text_blockchain)

print(f"AI Tokens ({len(tokens_ai)} tokens):")
print(tokens_ai)
print(f"\nBlockchain Tokens ({len(tokens_blockchain)} tokens):")
print(tokens_blockchain)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

set_ai = set(tokens_ai)
set_blockchain = set(tokens_blockchain)

intersection = set_ai.intersection(set_blockchain)
union = set_ai.union(set_blockchain)
jaccard_similarity = len(intersection) / len(union)

print(f"Common words (Intersection): {intersection}")
print(f"Total unique words across both (Union size): {len(union)}")
print(f"Jaccard Similarity: {jaccard_similarity:.4f}")

clean_str_ai = " ".join(tokens_ai)
clean_str_blockchain = " ".join(tokens_blockchain)

sim_tfidf = TfidfVectorizer()
tfidf_vectors = sim_tfidf.fit_transform([clean_str_ai, clean_str_blockchain])
cos_sim_matrix = cosine_similarity(tfidf_vectors)
cosine_sim = cos_sim_matrix[0][1]

print(f"Cosine Similarity (TF-IDF): {cosine_sim:.4f}")
```

## Question 5


```python
from textblob import TextBlob
from wordcloud import WordCloud

reviews = [
    "The noise-cancelling headphones offer breathtaking sound quality, superb comfort, and extraordinary battery life!",
    "Customer service was terrible, extremely rude, and completely unhelpful. I had an awful experience.",
    "The laptop arrived on time in standard cardboard packaging with a charging cable included.",
    "Outstanding smartwatch! The health monitoring sensors are remarkably precise and the display is gorgeous.",
    "Disappointing purchase. The build quality feels cheap and it stopped working after just three days.",
    "The tablet has 128GB of internal storage and supports expandable memory via microSD."
]

results = []
for rev in reviews:
    blob = TextBlob(rev)
    polarity = round(blob.sentiment.polarity, 4)
    subjectivity = round(blob.sentiment.subjectivity, 4)

    if polarity > 0.1:
        sentiment = "Positive"
    elif polarity < -0.1:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    results.append({
        'Review': rev,
        'Polarity': polarity,
        'Subjectivity': subjectivity,
        'Sentiment': sentiment
    })

reviews_df = pd.DataFrame(results)
print("Sentiment Analysis Results:")
display(reviews_df)
```

```python
plt.figure(figsize=(7, 4))
sentiment_counts = reviews_df['Sentiment'].value_counts()
colors = {'Positive': '#2ecc71', 'Negative': '#e74c3c', 'Neutral': '#95a5a6'}
plt.bar(sentiment_counts.index, sentiment_counts.values, color=[colors[k] for k in sentiment_counts.index])
plt.title("Review Sentiment Classification Distribution")
plt.xlabel("Sentiment Class")
plt.ylabel("Number of Reviews")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

positive_texts = " ".join(reviews_df[reviews_df['Sentiment'] == 'Positive']['Review'])

wordcloud = WordCloud(
    width=800,
    height=400,
    background_color='white',
    colormap='viridis',
    stopwords=set(stopwords.words('english')),
    min_font_size=10
).generate(positive_texts)

plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis("off")
plt.title("Word Cloud of Positive Reviews", fontsize=16, pad=15)
plt.show()
```

## Question 6


```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout

corpus_paragraph = """Cognitive computing simulates human thought processes in a computerized model.
It involves self-learning systems that use machine learning, natural language processing, and pattern recognition.
These intelligent systems mimic the way the human brain thinks, learns, and reacts to stimuli.
By analyzing vast amounts of unstructured data, cognitive models provide actionable insights.
They help humans make better decisions in complex and uncertain environments.
In healthcare and diagnostics, cognitive assistants empower physicians to deliver personalized patient care.
With rapid advancements in neural architectures, cognitive computing continues to reshape modern technology and society."""

print("Training Corpus Paragraph (~100 words):")
print(corpus_paragraph)

corpus_lines = [line.strip() for line in corpus_paragraph.strip().split('\n') if line.strip()]

tokenizer = Tokenizer()
tokenizer.fit_on_texts(corpus_lines)
total_words = len(tokenizer.word_index) + 1
print(f"\nTotal Unique Words in Vocabulary: {total_words}")
```

```python
input_sequences = []
for line in corpus_lines:
    token_list = tokenizer.texts_to_sequences([line])[0]
    for i in range(1, len(token_list)):
        n_gram_sequence = token_list[:i+1]
        input_sequences.append(n_gram_sequence)

max_sequence_len = max([len(seq) for seq in input_sequences])
input_sequences = np.array(pad_sequences(input_sequences, maxlen=max_sequence_len, padding='pre'))

X, y = input_sequences[:, :-1], input_sequences[:, -1]
y = tf.keras.utils.to_categorical(y, num_classes=total_words)

print(f"Total generated N-gram training sequences: {len(input_sequences)}")
print(f"Maximum sequence length: {max_sequence_len}")
print(f"Predictor feature matrix shape (X): {X.shape}")
print(f"Target one-hot label matrix shape (y): {y.shape}")
```

```python
model = Sequential([
    Embedding(total_words, 64),
    LSTM(100),
    Dropout(0.1),
    Dense(total_words, activation='softmax')
])

model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
model.summary()

history = model.fit(X, y, epochs=100, verbose=0)
print(f"Training Complete! Final Epoch Loss: {history.history['loss'][-1]:.4f}, Accuracy: {history.history['accuracy'][-1]:.4f}")
```

```python
def generate_text_lines(seed_text, next_words, num_lines=3):
    current_seed = seed_text
    generated_lines = []

    for line_idx in range(num_lines):
        line_words = []
        for _ in range(next_words):
            token_list = tokenizer.texts_to_sequences([current_seed])[0]
            token_list = pad_sequences([token_list], maxlen=max_sequence_len - 1, padding='pre')
            predicted_probs = model.predict(token_list, verbose=0)[0]
            predicted_idx = np.argmax(predicted_probs)

            output_word = ""
            for word, index in tokenizer.word_index.items():
                if index == predicted_idx:
                    output_word = word
                    break

            current_seed += " " + output_word
            line_words.append(output_word)

        generated_lines.append(" ".join(line_words))
    return generated_lines

seed_word = "cognitive"
print(f"Seed Word: '{seed_word}'\n")
generated_lines = generate_text_lines(seed_word, next_words=7, num_lines=3)

print("Generated Text (2-3 new lines):")
print("-" * 50)
for i, line in enumerate(generated_lines, start=1):
    print(f"Line {i}: {line}")
print("-" * 50)
```
