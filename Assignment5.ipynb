```python
import numpy as np
```

```python
arr = np.array([1,2,3,4,5])
print("Addition of 2")
print(arr+2)
print("Multiplication of 2")
print(arr*2)
print("Division by 2")
print(arr/2)
```

```python
arr = np.array([1,2,3,4,5,6])
reversed_arr = arr[::-1]

print("Original")
print(arr)
print("Reversed")
print(reversed_arr)
```

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
values, counts = np.unique(x, return_counts=True)
most_freq = values[np.argmax(counts)]
indices=np.where(x == most_freq)[0]
print("most freq value:",most_freq)
print("indices:",indices)
```

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
values,counts =np.unique(y,return_counts=True)
most_freq =values[np.argmax(counts)]
indices = np.where(y == most_freq)[0]
print("most frequent value:", most_freq)
print("indices:", indices)
```

```python
arr = np.array([[10, 20, 30],[40, 50, 60],[70, 80, 90]])
print("1st row, 2nd column:", arr[0, 1])
print("3rd row, 1st column:", arr[2, 0])
```

```python
gauransh = np.linspace(10, 100, 25)
print("Array:")
print(gauransh)
print("Dimensions:", gauransh.ndim)
print("Shape:", gauransh.shape)
print("Total elements:", gauransh.size)
print("Data type:", gauransh.dtype)
print("Total bytes:", gauransh.nbytes)
transposed = gauransh.reshape(1, -1)
print("Transpose using reshape():")
print(transposed)
print("Transpose using T:")
print(gauransh.T)
```

```python
ucs420_gauransh = np.array([[10, 20, 30, 40],[50, 60, 70, 80],[90, 15, 20, 35]])
print("Original array:")
print(ucs420_gauransh)
print("Mean:", np.mean(ucs420_gauransh))

print("Median:", np.median(ucs420_gauransh))

print("Maximum:", np.max(ucs420_gauransh))

print("Minimum:", np.min(ucs420_gauransh))

print("Unique elements:", np.unique(ucs420_gauransh))

reshaped_ucs420_gauransh = ucs420_gauransh.reshape(4, 3)
print("\nReshaped array:")
print(reshaped_ucs420_gauransh)

resized_ucs420_gauransh = np.resize(ucs420_gauransh, (2, 3))
print("\nResized array:")
print(resized_ucs420_gauransh)
```
